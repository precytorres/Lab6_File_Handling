```python
print("=== Most Frequent IP ===")
print(weblog['IP'].value_counts().head(5))

print("\n=== Most Visited URL ===")
print(weblog['URL'].value_counts().head(5))

print("\n=== Status Counts ===")
print(weblog['Staus'].value_counts())
```

```python
# Exercise 2: Online Sales Dataset
sales = pd.read_csv('data/Online Sales Data.csv')
print("Rows:", sales.shape)
print(sales.head())
print(sales.columns.tolist())
print(sales.info())
```

```python
print("=== Total Sales per Category ===")
print(sales.groupby('Product Category')['Total Revenue'].sum())

print("\n=== Total Sales per Region ===")
print(sales.groupby('Region')['Total Revenue'].sum())

print("\n=== Most Sold Product ===")
print(sales.groupby('Product Name')['Units Sold'].sum().sort_values(ascending=False).head(5))

print("\n=== Payment Method Counts ===")
print(sales['Payment Method'].value_counts())
```

```python
# Save Exercise 2 results
with open('output/sales_summary.txt', 'w') as f:
    f.write("ONLINE SALES ANALYSIS\n")
    f.write(f"Total Rows: {len(sales)}\n\n")
    f.write("Total Sales per Category:\n")
    f.write(sales.groupby('Product Category')['Total Revenue'].sum().to_string())
    f.write("\n\nTotal Sales per Region:\n")
    f.write(sales.groupby('Region')['Total Revenue'].sum().to_string())
    f.write("\n\nTop 5 Most Sold Products:\n")
    f.write(sales.groupby('Product Name')['Units Sold'].sum().sort_values(ascending=False).head(5).to_string())
    f.write("\n\nPayment Method Counts:\n")
    f.write(sales['Payment Method'].value_counts().to_string())

print("Saved! Check output/sales_summary.txt")
```

```python
# Exercise 3: Household Power Consumption
power = pd.read_csv('data/household_power_consumption.txt', sep=';', low_memory=False, na_values='?')
print("Rows:", power.shape)
print(power.head())
print(power.columns.tolist())
print(power.dtypes)
```

```python
# Convert to numeric
cols = ['Global_active_power', 'Global_reactive_power', 'Voltage', 'Global_intensity', 'Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3']
for c in cols:
    power[c] = pd.to_numeric(power[c], errors='coerce')

print("=== Missing Values ===")
print(power[cols].isna().sum())

print("\n=== Average Consumption ===")
print(power[cols].mean())

print("\n=== Max Voltage ===")
print("Max:", power['Voltage'].max())
print("Min:", power['Voltage'].min())
```

```python
# Save Exercise 3 results
with open('output/power_summary.txt', 'w') as f:
    f.write("HOUSEHOLD POWER CONSUMPTION ANALYSIS\n")
    f.write(f"Total Rows: {len(power)}\n\n")
    f.write("Missing Values:\n")
    f.write(power[cols].isna().sum().to_string())
    f.write("\n\nAverage Consumption:\n")
    f.write(power[cols].mean().to_string())
    f.write(f"\n\nMax Voltage: {power['Voltage'].max()}")
    f.write(f"\nMin Voltage: {power['Voltage'].min()}")

print("Saved! Check output/power_summary.txt")
```
